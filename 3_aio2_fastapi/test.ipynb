```python
class NormalUser:
    name: str = "krchoi"


user1 = NormalUser()
user1.name = 10
print(user1.name)
```

```python
# import pydantic
from pydantic import BaseModel


class PydanticUser(BaseModel):
    name: str


user2 = PydanticUser(name="10")

print(user2.name)
```

```python
class UserRegister(BaseModel):
    user_id: int
    user_name: str
    email: str
    is_admin: bool = False


user3 = UserRegister(
    user_id=5, user_name="krchoi", email="kyungrac12@gamil.com", is_admin=True
)

print(user3)
```

```python
input_data = {
    "user_id": "1004",  # 문자열로 들어왔지만 int로 자동 변환(Parsing)됩니다.
    "user_name": "codemaster",
    "email": "code@example.com",
    # is_admin은 생략되었으므로 기본값 False가 적용됩니다.
}

user1 = UserRegister(**input_data)

print(type(user1.user_id))


user1.model_dump
```
