# 몇상자를 주문하시겠어요? (숫자로)

# 빼고 싶은 간식이 있으면 번호를 입력하세요.

# 주문해 주셔서 감사합니다.

```python
snack_box = ['cake']

append = input("enter your favorite snack : ").split('/')
snack_box += append

print("\n현재 간식 목록")
for i, snack in enumerate(snack_box, start=1):
    print(f"{i}. {snack}")

# 1. 몇 상자를 주문하시겠어요?
box = int(input("\n몇 상자를 주문하시겠어요? (숫자로) : "))

# 2. 빼고 싶은 간식이 있으면 번호를 입력하세요.
remove_num = int(input("빼고 싶은 간식이 있으면 번호를 입력하세요 (없으면 0) : "))

if remove_num != 0:
    if 1 <= remove_num <= len(snack_box):
        removed = snack_box.pop(remove_num - 1)
        print(f"{removed}이(가) 삭제되었습니다.")
    else:
        print("잘못된 번호입니다.")

# 결과 출력
print("\n최종 주문 목록")
for snack in snack_box:
    print("-", snack)

print(f"주문 상자 수 : {box}")

# 3. 주문해 주셔서 감사합니다.
print("\n주문해 주셔서 감사합니다.")
```

# Check Sentence

```python
sentence = input("문장을 입력하세요: ")
words = ["파이썬", "코딩", "데이터", "AI"]
found_word = None
for word in words:
    if word in sentence:
        found_word = word
        break
if found_word:
     print(f"입력한 문장에서 '{found_word}' 단어를 찾았습니다!")
else:
     print("입력한 문장에 해당 단어가 없습니다.")
```
