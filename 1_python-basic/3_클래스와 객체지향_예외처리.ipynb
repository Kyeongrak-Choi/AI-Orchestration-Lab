## Q1. 다음과 같이 동작하는 클래스 Calculator를 작성해 보자.

>cal1 = Calculator([1,2,3,4,5])
>cal1.sum() # 15
>cal1.avg() # 3.0

>cal2 = Calculator([6,7,8,9,10])
>cal2.sum() # 40
>cal2.avg() #8.0

```python
class Calculator:
    def __init__(self,numbers):
        self.numbers = list(numbers)
    def sum(self):
        return sum(self.numbers)
    def avg(self):
        if not self.numbers:
            return 0
        return sum(self.numbers) / len(self.numbers)

cal1 = Calculator([1,2,3,4,5])
print(f"cal1.sum() : {cal1.sum()}")
print(f"cal1.avg() : {cal1.avg()}")

print('=' * 20)

cal2 = Calculator([6,7,8,9,10])
print(f"cal2.sum() : {cal2.sum()}")
print(f"cal2.avg() : {cal2.avg()}")
```

# 예외 클래스 정의

```python
class AgeException(Exception):
    def __init__(self,msg):
        self.message = msg
```

```python
def input_age():
    age = int(input("input your age"))

    # 입력체크
    if age <= 0:
        raise AgeException("나이는 0보다 커야 합니다")
    elif age > 100:
        raise AgeException("나이는 100보다 작아야 합니다")
    else:
        print(f"나이 : {age}")
```

```python
try:
    input_age()
except AgeException as ae:
    print(f"나이 범위는 0보다 커야 하고 100보다 작아야 합니다: {ae}")
```
