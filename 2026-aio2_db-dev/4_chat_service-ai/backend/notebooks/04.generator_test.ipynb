```python
def simple_generator():
    print("제너레이터 실행 시작")
    yield 1 # 첫 번째 next() 호출 시 여기까지 실행되고 1 반환 후 멈춤
    print("첫 번째 yield 이후 재개됨")
    yield 2 # 두 번째 next() 호출 시 여기까지 실행되고 2 반환 후 멈춤
    print("두 번째 yield 이후 재개됨")
    yield 3 # 세 번째 next() 호출 시 여기까지 실행되고 3 반환 후 멈춤
    print("제너레이터 실행 완료")
    # 함수가 끝나면 자동으로 StopIteration 발생

```

```python
import datetime # 시간 정보 활용
import time

print(f"--- 제너레이터 함수 동작 확인 (현재 시각: {datetime.datetime.now().strftime('%H:%M:%S KST')}) ---") # 오전 11:54

def timed_counter(limit):
    """1부터 limit까지 숫자를 1초 간격으로 생성하는 제너레이터"""
    print("  [G] 카운터 시작")
    n = 1
    while n <= limit:
        print(f"  [G] {n} 생성 준비...")
        yield n # n 값을 반환하고 여기서 멈춤
        print(f"  [G] {n} 반환 후 재개됨")
        time.sleep(1) # 1초 대기 (다음 next() 호출 때까지 영향 없음)
        n += 1
    print("  [G] 카운터 완료")

# 1. 제너레이터 객체 생성
counter_gen_obj = timed_counter(3)
print(f"생성된 객체 타입: {type(counter_gen_obj)}") # <class 'generator'>

# 2. 수동으로 next() 호출
print("\n수동 next() 호출:")
try:
    val1 = next(counter_gen_obj) # 함수 실행 시작 -> 첫 yield 까지
    print(f"  [C] 첫 번째 값 받음: {val1}")
    val2 = next(counter_gen_obj) # 함수 재개 -> 두 번째 yield 까지
    print(f"  [C] 두 번째 값 받음: {val2}")
    val3 = next(counter_gen_obj) # 함수 재개 -> 세 번째 yield 까지
    print(f"  [C] 세 번째 값 받음: {val3}")
    val4 = next(counter_gen_obj) # 함수 재개 -> 루프 종료 -> StopIteration 발생
except StopIteration:
    print("  [C] StopIteration 발생: 더 이상 값이 없음")

# 3. for 루프로 사용 (가장 일반적)
print("\nfor 루프로 사용:")
# 새 제너레이터 객체를 만들어야 함 (이전 것은 소진됨)
for number in timed_counter(2):
    print(f"  [F] for 루프에서 받은 값: {number}")
```

```python
조각 = ["안녕", "하세요. ", "면접을 ", "시작하죠."]

def 리턴판():
    모음 = ""
    for c in 조각:
        time.sleep(0.3)      # 모델이 조각을 만드는 시간
        모음 += c
    return 모음               # 다 만든 뒤에 한 번에 준다

def 이일드판():
    모음 = ""
    for c in 조각:
        time.sleep(0.3)
        모음 += c
        yield c               # 조각을 주고 그 자리에 멈춘다
```

```python
이일드판()

for val in 이일드판():
    print(val)
```
