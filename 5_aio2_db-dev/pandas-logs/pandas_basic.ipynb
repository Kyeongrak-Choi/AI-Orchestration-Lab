## 1. Series — 한 줄짜리 데이터

**Series** 는 값들이 한 줄로 늘어선 것입니다. 각 값에 이름표(인덱스)가 붙습니다.

```python
import pandas as pd

점수 = pd.Series([90, 85, 100], index=["국어", "영어", "수학"])
print(점수)
print()
print("수학 점수:", 점수["수학"])
```

딕셔너리로 만들면 키가 그대로 인덱스가 됩니다.

```python
# TODO 1. 딕셔너리로 Series 를 만든다.  힌트: pd.Series({"사과": 3000, ...})
가격 = pd.Series({"apple": 3000, "banana":5000,"grape":2000,"pear":1500})

print(가격)
print()
print("평균:", 가격.mean())
```

## 2. DataFrame — 표

**DataFrame** 은 Series 를 여러 개 옆으로 붙인 표입니다. 열마다 이름이 있습니다.

```python
data = {
    "이름": ["김철수", "이영희", "박민수", "최지우"],
    "직무": ["백엔드", "프런트", "백엔드", "데이터"],
    "경력": [3, 5, 1, 7],
}

df = pd.DataFrame(data)
df
```

새 열은 이름을 지어 값을 넣으면 만들어집니다. 다른 열로 계산해도 됩니다.

```python
# TODO 2. 경력이 1년 이하면 True 인 "신입" 열을 만든다.  힌트: df["경력"] <= 1
df["신입"] = df["경력"] <= 1

df
```

## 3. 골라내기

열 하나는 `df["열이름"]`, 여러 개는 대괄호를 두 번 씁니다.

```python
print(df["이름"])
print()
df[["이름", "경력"]]
```

행은 **조건**으로 고릅니다. 조건이 `True` 인 행만 남습니다.

```python
df[df["경력"] >= 3]
```

조건이 둘이면 `&`(그리고) 또는 `|`(또는) 로 잇고, **각 조건을 괄호로 묶습니다.**

괄호를 빠뜨리면 오류가 납니다. 파이썬이 `&` 를 먼저 계산하려 들기 때문입니다.

```python
# TODO 3. 경력 3년 이상이면서 직무가 백엔드인 행만 고른다.
#         힌트: df[(조건1) & (조건2)]

df[(df["경력"] >= 3) & (df["직무"] == "백엔드")]

```

반대로 고르려면 조건 앞에 `~` 를 붙입니다.

```python
df[~(df["직무"] == "백엔드")]
```

## 7. 묶어서 보기

`groupby` 는 **같은 값끼리 묶어서** 계산합니다. "메서드별 평균 응답시간" 처럼 나눠 볼 때 씁니다.

## 8. 빈 값

빈 값은 `NaN` 으로 들어옵니다. 먼저 **얼마나 있는지** 봅니다.

```python
logs.isna().sum()
```

처리 방법은 둘입니다.

| | 하는 일 | 언제 |
| --- | --- | --- |
| `fillna(값)` | 빈 자리를 값으로 채운다 | 0 이나 "없음" 으로 둬도 되는 경우 |
| `dropna(subset=[...])` | 그 열이 빈 행을 지운다 | 값이 없으면 계산이 안 되는 경우 |

**`dropna()` 를 그냥 쓰면 한 열이라도 비면 그 행이 사라집니다.** `subset` 으로 볼 열을 정해줍니다.

```python
print("원본:", len(logs))
print("dropna() 그냥  :", len(logs.dropna()))
print("dropna() SUBSET  :", len(logs.dropna(subset=["duration_ms"])))

# TODO 7. duration_ms 가 빈 행만 지웠을 때 몇 행이 남는지 센다.
#         힌트: logs.dropna(subset=["duration_ms"])
print("subset 지정    :", ...)
```

## 10. 날짜

글자로 읽힌 시각은 그대로는 날짜로 못 씁니다. `pd.to_datetime` 으로 바꾼 뒤 `.dt` 로 꺼냅니다.

```python
logs["시각"] = pd.to_datetime(logs["ts"])

print(logs["시각"].dt.date.head(3).to_string())
print()
print("시간대별 요청 수")
print(logs["시각"].dt.hour.value_counts().sort_index().to_string())
```
